```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from src.data_processor import TennisDataProcessor

# Load and process data
processor = TennisDataProcessor()
matches = processor.load_matches_data(data_path="../data/raw/")
clean_matches = processor.clean_matches_data()

# Test ELO system
elo_matches, current_ratings = processor.calculate_elo_ratings()

# Validation tests
print("=== TOP 20 PLAYERS BY ELO ===")
top_players = sorted(current_ratings.items(), key=lambda x: x[1], reverse=True)[:20]
for i, (player, rating) in enumerate(top_players, 1):
    print(f"{i:2d}. {player}: {rating:.0f}")

# Additional validation
print(f"\n=== ELO STATISTICS ===")
ratings_list = list(current_ratings.values())
print(f"Mean rating: {np.mean(ratings_list):.0f}")
print(f"Std deviation: {np.std(ratings_list):.0f}")
print(f"Min rating: {min(ratings_list):.0f}")
print(f"Max rating: {max(ratings_list):.0f}")

# Plot ELO distribution
plt.figure(figsize=(10, 6))
plt.hist(ratings_list, bins=50, alpha=0.7)
plt.xlabel('ELO Rating')
plt.ylabel('Number of Players')
plt.title('Distribution of Player ELO Ratings')
plt.show()
```
