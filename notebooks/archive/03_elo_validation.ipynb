```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from src.data_processor import TennisDataProcessor
```

```python
# Load and process data
processor = TennisDataProcessor()
matches = processor.load_matches_data(data_path="../data/raw/")
clean_matches = processor.clean_matches_data()
```

```python
# Test ELO system

print("=== GENERAL ELO SYSTEM VALIDATION ===")

elo_matches, current_ratings = processor.calculate_elo_ratings()

# Validation tests
print("=== TOP 20 PLAYERS BY ELO ===")
top_players = sorted(current_ratings.items(), key=lambda x: x[1], reverse=True)[:20]
for i, (player, rating) in enumerate(top_players, 1):
    print(f"{i:2d}. {player}: {rating:.0f}")

# Additional validation
print(f"\n=== ELO STATISTICS ===")
ratings_list = list(current_ratings.values())
print(f"Mean rating: {np.mean(ratings_list):.0f}")
print(f"Std deviation: {np.std(ratings_list):.0f}")
print(f"Min rating: {min(ratings_list):.0f}")
print(f"Max rating: {max(ratings_list):.0f}")

# Plot ELO distribution
plt.figure(figsize=(10, 6))
plt.hist(ratings_list, bins=50, alpha=0.7)
plt.xlabel('ELO Rating')
plt.ylabel('Number of Players')
plt.title('Distribution of Player ELO Ratings')
plt.show()
```

```python
# Test surface-specific ELO system

print("=== SURFACE-SPECIFIC ELO VALIDATION ===")

surface_ratings, surface_matches = processor.calculate_surface_elo()

# Surface-specific validation
for surface, ratings in surface_ratings.items():
    print(f"\n{surface} court: {len(ratings)} players")

    if ratings:
        top_on_surface = sorted(ratings.items(), key=lambda x: x[1], reverse=True)[:5]
        print(f"Top 5 players on {surface}:")
        for i, (player, rating) in enumerate(top_on_surface, 1):
            print(f" {i}. {player}: {rating:.0f}")

# Surface-specific statistics
print(f"\n=== SURFACE-SPECIFIC ELO STATISTICS ===")

for surface, ratings in surface_ratings.items():
    ratings_list = list(ratings.values())
    print(f"\n{surface} Court ({len(ratings)} players):")
    print(f"  Mean: {np.mean(ratings_list):.0f}")
    print(f"  Std Dev: {np.std(ratings_list):.0f}")
    print(f"  Range: {min(ratings_list):.0f} - {max(ratings_list):.0f}")
```
